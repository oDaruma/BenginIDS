## Interpretation checklist
1. Rank candidates by PR-AUC, then inspect uncertainty across folds/seeds.
2. Compare precision and recall at the validation-selected $\tau$.
3. Check whether optimization gains justify computation time.
4. Inspect category-level recall before deployment.
5. Treat the test result as evidence for this split—not a universal claim.

# 4. Thresholds, interpretation, and robustness
A SOC deploys a threshold $\tau$ that trades analyst workload against missed attacks.

```python
import numpy as np
from benignids.evaluation import choose_threshold
y_val = np.array([0, 0, 0, 1, 1, 1])
y_score = np.array([0.05, 0.25, 0.55, 0.60, 0.80, 0.95])
tau_f1 = choose_threshold(y_val, y_score, objective='f1')
tau_high_precision = choose_threshold(y_val, y_score, minimum_precision=0.90)
tau_f1, tau_high_precision
```

## Required analyses
- PR curve and confusion matrix at the selected $\tau$
- Brier score and calibration
- SHAP for the tree baseline; permutation and masking ablations for the transformer
- repeated label-noise experiments with confidence intervals
- attack-category recall and drift across capture/time groups
